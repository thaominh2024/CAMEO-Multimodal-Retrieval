```python

from llama_cpp import Llama
import json
import pandas as pd

# Tải model từ Hugging Face
llm = Llama.from_pretrained(
    repo_id="Qwen/Qwen2.5-3B-Instruct-GGUF",
    filename="*q4_k_m.gguf",
    verbose=False,
    n_ctx=2048, # Độ dài ngữ cảnh
    repeat_penalty=1.2 # Thêm cái này để chặn lặp từ
)
```

```python
import re
def extract_product_info(text):
    prompt = f"""You are an expert e-commerce data extraction AI. Read the PRODUCT NAME and output a JSON object.

RULES:
1. NO HALLUCINATIONS. Extract exact words present in the text.
2. If information is missing, output an empty string "".

KEY DEFINITIONS:
- "ingredient": The main active ingredients, natural extracts, plants, herbs, or special materials (e.g., Aloe Vera, Centella, Mint, Red Ginseng, Rice milk, Snail, Tea Tree). DO NOT extract adjectives (Gentle, Purifying), product formats (Gel, Foam), or skin issues (Acne, Oily, Dirt).
- "capacity": Volume or weight containing numbers and units (e.g., 150ml, 591 ml, 50ml, 130ml, 80g).
- "type": The specific format of the product (e.g., Facial Cleanser, Foam Cleanser, Gel Cleanser, Face Wash).

--- EXAMPLES ---
Text: "Civic Rice milk Rice Facial Cleanser"
JSON: {{"ingredient": "Rice milk", "capacity": "", "type": "Facial Cleanser"}}

Text: "Centella and Mint Extract Facial Cleanser "
JSON: {{"ingredient": "Centella, Mint", "capacity": "50ml", "type": "Gel Cleanser"}}

Text: "Cetaphil Facial Cleanser 591 ml (Guaranteed 100%)"
JSON: {{"ingredient": "", "capacity": "591 ml", "type": "Facial Cleanser"}}
--- END OF EXAMPLES ---

PRODUCT NAME: "{text}"
JSON:
"""
    
    # Gọi model
    response = llm.create_chat_completion(
        messages=[{"role": "user", "content": prompt}],
        temperature=0.01
    )

    raw_content = response['choices'][0]['message']['content']

    try:
        # Tìm nội dung nằm giữa { và }
        match = re.search(r'\{.*\}', raw_content, re.DOTALL)
        if match:
            clean_json = match.group()
            return json.loads(clean_json)

        # Nếu AI không trả về ngoặc nhọn, tạo N/A mặc định
        return {k: "N/A" for k in ["ingredient", "capacity", "type"]}
    except:
        return {k: "N/A" for k in ["ingredient", "capacity", "type"]}
```

```python
import os
import pandas as pd

# 1. Đường dẫn chuẩn của Kaggle
output_csv_path = "/kaggle/working/Extracted_Products_Final.csv" 
chunk_size = 10
total_rows = 500

print(f"🚀 Bắt đầu xử lý tổng cộng {total_rows} sản phẩm trên Kaggle...")

for i in range(3241, total_rows+3241, chunk_size):
    chunk = df.iloc[i:i+chunk_size].copy()
    
    end_idx = min(i + chunk_size, total_rows)
    
    # Chạy AI trích xuất
    extracted_data = chunk['product_name'].apply(extract_product_info)
    
    # Tách dict thành 3 cột mới
    new_columns = pd.DataFrame(extracted_data.tolist(), index=chunk.index)
    chunk[['ingredient', 'capacity', 'type']] = new_columns[['ingredient', 'capacity', 'type']]
    
    if i == 0:
        chunk.to_csv(output_csv_path, mode='w', header=True, index=False, encoding='utf-8-sig')
    else:
        chunk.to_csv(output_csv_path, mode='a', header=False, index=False, encoding='utf-8-sig')
        
    print(f"✅ Đã chốt an toàn vào /kaggle/working/!")

print("🎉 TUYỆT VỜI! Đã xử lý xong.")
```
